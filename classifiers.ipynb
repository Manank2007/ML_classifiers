```python
from sklearn.datasets import make_moons
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

X, y = make_moons(n_samples=100, noise=0.1, random_state=42)
plt.figure(figsize=(6, 5))
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm', edgecolors='k', s=40)
plt.title("make_moons Dataset")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
print(y)
print(X)
print(X.shape)
print(y.shape)
```

```python
#as this is a non linearly seperable dataset , we will try to apply different classification algorithims to see how they perform on this dataset.

#SVM classifier with kernel function
from sklearn import svm
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
svm=svm.SVC(kernel='rbf', C=1.0, gamma=0.7)
model=svm.fit(X_train,y_train)
prediction=model.predict(X_test)
accuracy_score(y_test,prediction,normalize=True)

```

```python
y_predicted=model.predict(X_test)
print(y_predicted)
print(y_test)
plt.figure(figsize=(6, 5))
plt.scatter(X_test[:, 0], X_test[:, 1], c=y_predicted, cmap='coolwarm', edgecolors='k', s=40)
```

```python
# 1. Find the misclassified points 
misclassified_mask = (y_test != y_predicted)

# Find the coordinates of the wrong point
x_wrong = X_test[misclassified_mask, 0]
y_wrong = X_test[misclassified_mask, 1]
```

```python
figure,axes=plt.subplots(1,2,figsize=(12,5))
axes[0].scatter(X_test[:, 0], X_test[:, 1], c=y_predicted, cmap='coolwarm', edgecolors='k', s=40)
axes[0].scatter(x_wrong, y_wrong, facecolors='none', edgecolors='gold', 
            linewidths=3, s=250, label='Misclassified')
axes[1].scatter(X_test[:, 0], X_test[:, 1], c=y_test, cmap='coolwarm', edgecolors='k', s=40)
axes[1].scatter(x_wrong, y_wrong, facecolors='none', edgecolors='gold', 
            linewidths=3, s=250, label='Misclassified')
axes[1].set_title("Actual Labels")
axes[0].set_title("Predicted Labels")
plt.legend()
```

```python
#Using K-nearest Neighbour classifier(K-NN)

from sklearn.neighbors import KNeighborsClassifier
nn=KNeighborsClassifier(n_neighbors=3)
model=nn.fit(X_train,y_train)
nn.score(X_test,y_test)
y_pred_n=model.predict(X_test)
figure,axes=plt.subplots(1,2,figsize=(12,5))
axes[0].scatter(X_test[:,0],X_test[:,1],c=y_pred_n,cmap='coolwarm',edgecolors='k',s=40)
axes[1].scatter(X_test[:,0],X_test[:,1],c=y_test,cmap='coolwarm',edgecolors='k',s=40)
axes[1].set_title("Actual Labels")
axes[0].set_title("Predicted Labels")
```

```python
#Using Decison Tree Classifier
from sklearn.tree import DecisionTreeClassifier
dTree_entropy=DecisionTreeClassifier(criterion='entropy',random_state=100,max_depth=3,min_samples_leaf=5)
model=dTree_entropy.fit(X_train,y_train)
y_pred_DTE=model.predict(X_test)
accuracy_score(y_test,y_pred_DTE,normalize=True)

```

```python
#Decison Tree Classifier prediction plot

misclassified_mask_dte = (y_test != y_pred_DTE)

# Find the coordinates of the wrong point
x_wrong_dte = X_test[misclassified_mask_dte, 0]
y_wrong_dte = X_test[misclassified_mask_dte, 1]
figure,axes=plt.subplots(1,2,figsize=(12,5))
axes[0].scatter(X_test[:,0],X_test[:,1],c=y_pred_DTE,cmap='coolwarm',edgecolors='k',s=40)
axes[0].scatter(x_wrong_dte, y_wrong_dte, facecolors='none', edgecolors='gold', 
            linewidths=3, s=250, label='Misclassified')
axes[1].scatter(X_test[:,0],X_test[:,1],c=y_test,cmap='coolwarm',edgecolors='k',s=40)
axes[1].scatter(x_wrong_dte, y_wrong_dte, facecolors='none', edgecolors='gold', 
            linewidths=3, s=250, label='Misclassified')
axes[1].set_title("Actual Labels")
axes[0].set_title("Predicted Labels")
plt.legend()

```
