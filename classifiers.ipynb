```python
from sklearn.datasets import make_moons
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

X, y = make_moons(n_samples=100, noise=0.1, random_state=42)
plt.figure(figsize=(6, 5))
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm', edgecolors='k', s=40)
plt.title("make_moons Dataset")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
print(y)
print(X)
print(X.shape)
print(y.shape)
```

```python
#as this is a non linearly seperable dataset , we will try to apply different classification algorithims to see how they perform on this dataset.

#SVM classifier with kernel function
from sklearn import svm
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
import numpy as np
import matplotlib.pyplot as plt


def plot_decision_boundary(model, X_train, y_train, X_test, y_test, title):
    
    # -------------------------------------------------
    # Create a grid over the feature space
    # -------------------------------------------------
    x_min = X_train[:, 0].min() - 0.3
    x_max = X_train[:, 0].max() + 0.3

    y_min = X_train[:, 1].min() - 0.3
    y_max = X_train[:, 1].max() + 0.3

    xx, yy = np.meshgrid(
        np.linspace(x_min, x_max, 500),
        np.linspace(y_min, y_max, 500)
    )

    # Convert grid into points
    grid_points = np.c_[xx.ravel(), yy.ravel()]

    # Predict class for every point in the grid
    Z = model.predict(grid_points)

    # Reshape predictions back into grid shape
    Z = Z.reshape(xx.shape)

    # -------------------------------------------------
    # Create figure
    # -------------------------------------------------
    fig, ax = plt.subplots(figsize=(8, 6))

    # -------------------------------------------------
    # Decision regions
    # -------------------------------------------------
    ax.contourf(
        xx,
        yy,
        Z,
        alpha=0.18,
        cmap="coolwarm"
    )

    # -------------------------------------------------
    # Decision boundary
    # -------------------------------------------------
    ax.contour(
        xx,
        yy,
        Z,
        levels=[0.5],
        linewidths=2
    )

    # -------------------------------------------------
    # Training data - faded
    # -------------------------------------------------
    ax.scatter(
        X_train[:, 0],
        X_train[:, 1],
        c=y_train,
        cmap="coolwarm",
        alpha=0.25,
        s=35,
        edgecolors="none",
        label="Training data"
    )

    # -------------------------------------------------
    # Test data - prominent
    # -------------------------------------------------
    ax.scatter(
        X_test[:, 0],
        X_test[:, 1],
        c=y_test,
        cmap="coolwarm",
        s=55,
        edgecolors="black",
        linewidths=0.7,
        label="Test data"
    )

    # -------------------------------------------------
    # Find misclassified test points
    # -------------------------------------------------
    y_pred = model.predict(X_test)

    misclassified = y_pred != y_test

    ax.scatter(
        X_test[misclassified, 0],
        X_test[misclassified, 1],
        facecolors="none",
        edgecolors="black",
        linewidths=2,
        s=150,
        label="Misclassified"
    )

    # -------------------------------------------------
    # Formatting
    # -------------------------------------------------
    ax.set_title(
        title,
        fontsize=16,
        fontweight="bold",
        pad=12
    )

    ax.set_xlabel(
        "Feature 1",
        fontsize=12
    )

    ax.set_ylabel(
        "Feature 2",
        fontsize=12
    )

    ax.tick_params(
        labelsize=10
    )

    ax.grid(
        alpha=0.15,
        linestyle="--"
    )

    ax.legend(
        frameon=True,
        fontsize=10
    )

    # Keep identical axis limits
    ax.set_xlim(x_min, x_max)
    ax.set_ylim(y_min, y_max)

    plt.tight_layout()
    plt.show()
```

```python
svm=svm.SVC(kernel='rbf', C=1.0, gamma=0.7)
svm_model=svm.fit(X_train,y_train)
prediction=svm_model.predict(X_test)
accuracy_score(y_test,prediction,normalize=True)

```

```python
y_predicted=svm_model.predict(X_test)
print(y_predicted)
print(y_test)
plot_decision_boundary(
    svm_model,
    X_train,
    y_train,
    X_test,
    y_test,
    "SVM — RBF Kernel"
)
```

```python
#Using K-nearest Neighbour classifier(K-NN)

from sklearn.neighbors import KNeighborsClassifier
nn=KNeighborsClassifier(n_neighbors=3)
knn_model=nn.fit(X_train,y_train)
nn.score(X_test,y_test)
y_pred_n=knn_model.predict(X_test)
```

```python
plot_decision_boundary(
    knn_model,
    X_train,
    y_train,
    X_test,
    y_test,
    "K-Nearest Neighbours (K=3)"
)
```

```python
#Using Decison Tree Classifier
from sklearn.tree import DecisionTreeClassifier
dTree_entropy=DecisionTreeClassifier(criterion='entropy',random_state=100,max_depth=3,min_samples_leaf=5)
dt_model=dTree_entropy.fit(X_train,y_train)
y_pred_DTE=dt_model.predict(X_test)
accuracy_score(y_test,y_pred_DTE,normalize=True)

```

```python
#Decison Tree Classifier prediction plot



plot_decision_boundary(
    dt_model,
    X_train,
    y_train,
    X_test,
    y_test,
    "Decision Tree"
)
```

```python
#Random Forest Classifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix
import seaborn as sns
rf=RandomForestClassifier()
rf_model=rf.fit(X_train,y_train)
y_pred_rf=rf_model.predict(X_test)
accuracy_score(y_test,y_pred_rf,normalize=True)
```

```python
confusion_matrix(y_test,y_pred_rf)
sns.heatmap(confusion_matrix(y_test,y_pred_rf),annot=True,fmt='d',cmap='Blues')
```

```python
plot_decision_boundary(
    rf_model,
    X_train,
    y_train,
    X_test,
    y_test,
    "Random Forest"
)
```

```python
# Naive - Bayes Classifier
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import cross_val_score

gnb=GaussianNB()
nb_model=gnb.fit(X_train,y_train)
y_pred_nb=nb_model.predict(X_test)
scores = cross_val_score(nb_model, X, y, cv=5)
print(f"{scores.mean():.2%}")
```

```python
plot_decision_boundary(
    nb_model,
    X_train,
    y_train,
    X_test,
    y_test,
    "Gaussian Naive Bayes"
)
```
